```python
import pandas as pd
```

```python
# Extracting the data from the files

BSID = pd.read_csv('./cold_data/BSID_SERBIA_EXT.csv')
BSEG = pd.read_excel('./cold_data/BSEG_SERBIA_AROI.xlsx',header=0)
BKPF = pd.read_csv('./cold_data/BKPF_SERBIA_AROI.csv')
```

```python
# Cleaning the date by replacing '1900-01-01' with None 

BSID=BSID.replace('1900-01-01',None)
BSEG=BSEG.replace('1900-01-01',None)
BKPF=BKPF.replace('1900-01-01',None)
```
